```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
web = requests.get("https://www.iplt20.com/")
web
```

```python
soup= BeautifulSoup(web.content, "html.parser")
soup
```

```python
sections=soup.find_all("section",class_="flex w-full flex-col gap-y-6 py-5")
sections
```

```python
top_performer=sections[1]
top_performer
```

```python
heading=top_performer.find("div",class_="text-heading-5")
heading.text
```

```python
cards=top_performer.find_all("div", class_="relative h-[216px] cursor-default rounded-200 flex items-center justify-center overflow-hidden")
cards
```

```python
data=[]
for card in cards:
    images=card.find("img",{"alt":True})
    print(images["alt"])
    data.append(images["alt"])
```

```python
data
```

```python
data.remove('Fair Play Award, Punjab Kings')
data.remove('Vaibhav Sooryavanshi, Most Valuable Player, Matches 16, Points 436.55')
```

```python
names=[]
title=[]
matches=[]
performance=[]
point=[]
for i in data:
    list_of_data=i.split(",")
    names.append(list_of_data[0])
    title.append(list_of_data[1])
    matches.append(list_of_data[2])
    performance.append(list_of_data[3])
    point.append(list_of_data[4])
```

```python
data={
    "Name":names,
    "Title":title,
    "Matches":matches,
    "performance":performance,
    "Points":point
}
data
```

```python
df=pd.DataFrame(data)
df
```

```python
df.to_csv("player_dataset.csv",index=False)
```
