the above mentioned techniques are voting ensamble methods

Types of EL:

voting ensamble   different base model, 

stacking   train model on different model then train meta model on the output of the different model, complex and less used

in stacking output of different models are used to train a meta model and the output of the meta model is the final output

Bagging -> bootstrapped aggregation

base models can be same or different, use same model train on different data

sampling is done and is given to the model for example there are 1000 rows and 10 columns each model is given 500 rows and 5columns which are called samples

sampling types:

row sampling -> only rows are sampled     400 rows and 10 columns

column sampling -> only columns are sampled   1000 rows and 5 columns

hybrid sampling or mixed sampling -> both rows and columns are sampled   500 rows and 5 columns


all the sampling are further into

with replacement-> same data are available for other models

without replacement -> same data arent available for other models

all the models are trained parallelly, time is reduced so bagging is faster
