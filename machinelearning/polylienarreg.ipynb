polynomial lienar regression

transforms the features into polynomial feature where the feature are made x, x^2, x^3 for x degree of 3

simple polynomial reg -> only one input column

multiple polynomial reg -> many input columns

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from sklearn.preprocessing import PolynomialFeatures
```

```python
numbers=np.random.rand(100,1)
numbers.shape
```

```python
X=6*numbers-3
y=0.5*X**2+1.5*X+2+2*(np.random.rand(100,1))
```

```python
print(X)
```

```python
y
```

```python
plt.scatter(X,y)
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,y,test_size=0.3,random_state=41)
```

```python
lin_reg=LinearRegression()
lin_reg
```

```python
lin_reg.fit(X_train,Y_train)
lin_reg
```

```python
Y_pred=lin_reg.predict(X_test)
Y_pred
```

```python
score=r2_score(Y_test,Y_pred)
score
```

```python
plt.plot(X_train,lin_reg.predict(X_train),color='r')
plt.scatter(X_train,Y_train,color='g')
plt.xlabel('x')
plt.ylabel('y')
```

```python
poly=PolynomialFeatures(degree=2,include_bias=False)
X_train_poly=poly.fit_transform(X_train)
X_test_poly=poly.transform(X_test)
```

pipeline is sequence of data procressing and model training steps arranged in order, to automate and reproduce the entire ML work flow easily

it is simply connecting multiple ML steps into one continuous process

```python
from sklearn.pipeline import Pipeline
```

```python
def poly_regression(degree):
    X_new=np.linspace(-3,3,200).reshape(200,1)
    poly_features=PolynomialFeatures(degree=degree)
    lin_reg=LinearRegression()
    poly_regression=Pipeline([
        ('poly_features',poly_features),
        ('lin_reg',lin_reg)
    ])
    poly_regression.fit(X_train,Y_train)
    Y_pred_new=poly_regression.predict(X_new)
    plt.plot(X_new,Y_pred_new,label="predictions",color='r')
    plt.plot(X_train,Y_train,'g.',label='training data')
    plt.legend(loc="upper left")
    plt.xlabel('X dataset')
    plt.ylabel('y dataset')
    plt.show
```

```python
poly_regression(5)
```
