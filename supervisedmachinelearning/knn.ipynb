KNN classifier

```python
import pandas as pd
import seaborn as sns 
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification, make_regression
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.neighbors import KNeighborsClassifier, KNeighborsRegressor
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report,mean_squared_error,r2_score,mean_absolute_error
```

```python
X,Y=make_classification(
    n_samples=1000,
    n_features=2,
    n_redundant=0,
    n_classes=2,
    random_state=999
)
```

```python
X
```

```python
Y
```

```python
X.shape,Y.shape
```

```python
plt.scatter(X[:,0],X[:,1],c=Y)
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y, test_size=0.3, random_state=43)
```

```python
model=KNeighborsClassifier(metric='euclidean')
model
```

```python
model.fit(X_train,Y_train)
```

```python
Y_pred=model.predict(X_test)
```

```python
print(confusion_matrix(Y_test,Y_pred))
print(accuracy_score(Y_test,Y_pred))
print(classification_report(Y_test,Y_pred))
```

```python
parameters={
    "n_neighbors":[3,5,7,9,11,13,15],
    "metric":['euclidean','manhattan']
}
```

```python
clf=GridSearchCV(model,parameters)
```

```python
clf.fit(X_train,Y_train)
clf.best_params_
```

```python
Y_pred=clf.predict(X_test)
```

```python
print(confusion_matrix(Y_test,Y_pred))
print(accuracy_score(Y_test,Y_pred))
print(classification_report(Y_test,Y_pred))
```

```python
model=KNeighborsClassifier(metric='euclidean',n_neighbors=3)
```

```python
model.fit(X_train,Y_train)
```

```python
Y_pred=model.predict(X_test)
```

```python
print(confusion_matrix(Y_test,Y_pred))
print(accuracy_score(model.predict(X_train),Y_train))
print(accuracy_score(Y_test,Y_pred))
print(classification_report(Y_test,Y_pred))
```

KNN Regressor

```python
X,Y=make_regression(
    n_samples=1000,
    n_features=2,
    noise=10,
    random_state=43
)
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.3,random_state=43)
```

```python
model=KNeighborsRegressor(n_neighbors=5,algorithm='auto')
model
```

```python
model.fit(X_train,Y_train)
```

```python
Y_pred=model.predict(X_test)
```

```python
print(mean_absolute_error(Y_test,Y_pred))
print(mean_squared_error(Y_test,Y_pred))
print(r2_score(Y_test,Y_pred))
```
