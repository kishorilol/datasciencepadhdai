we pass the output of a linear regression to a sigmoid function

```python
from sklearn.datasets import load_iris
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split, GridSearchCV, RandomizedSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
import warnings 
warnings.filterwarnings('ignore')
import pickle
```

```python
dataset=load_iris()
```

```python
dataset
```

```python
print(dataset.DESCR)
```

```python
dataset.keys()
```

```python
dataset.feature_names
```

```python
df=pd.DataFrame(dataset.data,columns=dataset.feature_names)
```

```python
df.head()
```

```python
df['target']=dataset.target
```

```python
df.sample(10)
```

```python
df['target'].value_counts()
```

For Biclass

```python
df_copy=df[df['target']!=2]
```

```python
df_copy.sample(10)
```

```python
df_copy['target'].value_counts()
```

```python
X=df_copy.iloc[:,:-1]
Y=df_copy.iloc[:,-1]
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(X,Y,test_size=0.33,random_state=41)
```

```python
X_train
```

```python
classifier=LogisticRegression()
classifier
```

```python
classifier.fit(X_train,Y_train)
```

```python
Y_pred=classifier.predict(X_test)
```

```python
Y_pred
```

```python
classifier.predict_proba(X_test)
```

```python
print(confusion_matrix(Y_test,Y_pred))
```

```python
print(accuracy_score(Y_test,Y_pred))
```

```python
print(classification_report(Y_test,Y_pred))
```

Hyperparameter tuning

```python
parameter={
    'penalty':['l1','l2','elasticnet',None],
    'dual':[True,False],
    'C':[1,10,20,30,40,50],
    'max_iter':[100,200,300,1000,10000]
}
```

```python
clf=GridSearchCV(classifier,param_grid=parameter)
```

```python
clf
```

```python
clf.fit(X_train,Y_train)
```

```python
clf.best_params_
```

```python
clf.predict(X_test)
```

```python
classifier=LogisticRegression(C=1,penalty='l2',max_iter=100,dual=False)
```

```python
classifier.fit(X_train,Y_train)
```

```python
Y_pred=classifier.predict(X_test)
```

```python
print(confusion_matrix(Y_test,Y_pred))
print(accuracy_score(Y_test,Y_pred))
print(classification_report(Y_test,Y_pred))
```

Randomized Search CV

```python
abc=LogisticRegression()
```

```python
r_clf=RandomizedSearchCV(abc,param_distributions=parameter)
r_clf
```

```python
r_clf.fit(X_train,Y_train)
```

```python
r_clf.best_params_
```

```python
classifier=LogisticRegression(C=20,penalty='l2',max_iter=1000)
```

```python
classifier.fit(X_train,Y_train)
```

```python
Y_pred=classifier.predict(X_test)
```

```python
print(confusion_matrix(Y_test,Y_pred))
print(accuracy_score(Y_test,Y_pred))
print(classification_report(Y_test,Y_pred))
```

```python
pickle_file='logistic.pkl'
with open(pickle_file,'wb') as file:
    pickle.dump(classifier,file)
print(f'model saved to {pickle_file}')
```

```python
with open(pickle_file,'rb')as file:
    loaded_model=pickle.load(file)
```

```python
Y_pred_loaded=loaded_model.predict(X_test)
print(Y_pred_loaded)
```
