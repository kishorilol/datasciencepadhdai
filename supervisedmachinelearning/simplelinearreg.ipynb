```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
```

```python
df=pd.read_csv('height-weight.csv',index_col=0)
df.sample(10)
```

```python
df.shape
```

```python
plt.scatter(data=df, x='Height', y='Weight')
plt.xlabel('Height')
plt.ylabel('Weight')
plt.title('Height vs Weight')
```

```python
X=df[['Height']]
y=df[['Weight']]
```

```python
X,y
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y, test_size=0.20, random_state=41)
```

```python
df.info()
```

```python
df.describe()
```

```python
scaler=StandardScaler()
```

```python
np.mean(X_train),np.std(X_train), np.mean(X_test),np.std(X_test)
```

```python
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)
```

```python
np.mean(X_train),np.std(X_train), np.mean(X_test),np.std(X_test)
```

```python
plt.scatter(X_train,y_train)
```

```python
model=LinearRegression()
model
```

```python
model.fit(X_train,y_train)
```

```python
print(model.coef_)
print(model.intercept_)
```

```python
plt.scatter(X_train,y_train)
plt.plot(X_train,model.predict(X_train),'r')
```

```python
y_predict_test=model.predict(X_test)
```

```python
y_predict_test,y_test
```

```python
plt.scatter(X_test,y_test)
plt.plot(X_test,model.predict(X_test),'r')
```

```python
mse=mean_squared_error(y_test,y_predict_test)
mae=mean_absolute_error(y_test,y_predict_test)
rmse=np.sqrt(mse)
score=r2_score(y_test,y_predict_test)
print(mse,mae,rmse,score*100,'%')
```
