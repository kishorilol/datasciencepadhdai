2. creating array from scratch
. np.zeros() and np.ones() to create array of 0s and 1s
. np.arange() and np.linspace() to generate sequences of numbers
. np.random.rand() and np.random and np.random.radient() for random numbers

```python
print(np.zeros(5))
print(np.ones(5))
print(np.arange(5.5,10.5,0.5))
print(np.linspace(1,10,20,endpoint=False))
print(np.empty(5)) # Creates garbage values in the array, as it does not initialize the array elements.
print(np.random.rand(5)) # Creates an array of 5 random numbers between 0 and 1.
print(np.random.randint(1,10,5)) # Creates an array of 5 random integers between 1 and 10.
```

note: 
. arange accepts decimal value unlike range
. linspace takes third parameter as number of elements, and end value is inclusive unless endpoint=False
. end value are exclusive in random
. if no number of values assigend then it gives only one

operations in array
vector wise operation is done

```python
list1=[1,2,3,4,5]
list2=[6,7,8,9,10]
array1=np.array(list1)
array2=np.array(list2)
print(list1+list2)
print(array1+array2) # Adds the corresponding elements of the two arrays.
print(array1*2)
```

array indexing

```python
list1=[i+3 for i in range(50)]
print(list1)
array1=np.array(list1)
print(array1[0])
print(array1[-1])
print(array1[10:25:2])
print(array1[-1:-10:-1])
```
