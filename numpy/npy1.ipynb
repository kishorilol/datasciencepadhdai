2. creating array from scratch
. np.zeros() and np.ones() to create array of 0s and 1s
. np.arange() and np.linspace() to generate sequences of numbers
. np.random.rand() and np.random and np.random.radient() for random numbers

```python
print(np.zeros(5))
print(np.ones(5))
print(np.arange(5.5,10.5,0.5))
print(np.linspace(1,10,20,endpoint=False))
print(np.empty(5)) # Creates garbage values in the array, as it does not initialize the array elements.
print(np.random.rand(5)) # Creates an array of 5 random numbers between 0 and 1.
print(np.random.randint(1,10,5)) # Creates an array of 5 random integers between 1 and 10.
```

fancy indexing => access multiple index at same time  

```python
arr=np.random.randint(1,100,20)
print(arr)
#arr[2,-1,-6] error normal indexing
arr[[2,-1,-6]] #fancy indexing use [[]]
```

```python
arr=np.array([5,10,15,20,25,30,35,40,45,50])
print(arr)
print(arr[1::3]) #1. slice and print every third element from index 1
print(arr[-3::]) #2. last 4 element using negative indexing
print(arr[[2,4,6,8]]) #4. use fancy indexing to extract the values at positions [2,4,6,8]
arr[::2]=-1 #3. Replace the values at even indices with -1
print(arr)
```

array attributes and mathematical operations
1. ndim => gives dimension
2. shape => size of the array in each dimension
3. size => total number of element in array
4. dtype => type of data stored in the array
5. itemsize => size of each element in bytes
6. nbytes => total memory used by the array (itemsize x size)

```python
array=np.array([i for i in range(1,10)], dtype=np.float64)
print(array.nbytes)
print(array.itemsize)
print(array.dtype)
print(array.ndim)
print(array.shape)
print(array.size)
```

mathematical operation

```python
list1=[1,2,3,4,5]
list2=[6,7,8,9,10]
array1=np.array(list1)
array2=np.array(list2)
print(array1+array2)
print(np.add(array1,array2))
print(np.subtract(array1,array2))
print(np.multiply(array1,array2))
print(np.power(array1,array2))
print(np.divide(array1,array2))
```

logical operators

```python
x=np.array([True,True,False,False])
y=np.array([True,False,True,False])
print(x|y)
print(x&y)
print(~x)
print(np.logical_or(x,y))
print(np.logical_and(x,y))
print(np.logical_not(x))
```

statistical operators
no mode need to use scipy import stats

```python
from scipy import stats
arr=np.array([10,21,32,12,13,18,10,34,12,10])
print(np.median(arr))
# mode=stats.mode(arr)
# print(mode.mode,mode.count)
print(np.std(arr))
print(np.var(arr))
print(np.percentile(arr,80)) # values form 0 - 100
print(np.quantile(arr,0.80)) # values form 0-1
print(np.min(arr))
print(np.max(arr))
```
