2. creating array from scratch
. np.zeros() and np.ones() to create array of 0s and 1s
. np.arange() and np.linspace() to generate sequences of numbers
. np.random.rand() and np.random and np.random.radient() for random numbers

```python
print(np.zeros(5))
print(np.ones(5))
print(np.arange(5.5,10.5,0.5))
print(np.linspace(1,10,20,endpoint=False))
print(np.empty(5)) # Creates garbage values in the array, as it does not initialize the array elements.
print(np.random.rand(5)) # Creates an array of 5 random numbers between 0 and 1.
print(np.random.randint(1,10,5)) # Creates an array of 5 random integers between 1 and 10.
```

note: 
. arange accepts decimal value unlike range
. linspace takes third parameter as number of elements, and end value is inclusive unless endpoint=False
. end value are exclusive in random

operations in array
vector wise operation is done

```python
list1=[1,2,3,4,5]
list2=[6,7,8,9,10]
array1=np.array(list1)
array2=np.array(list2)
print(list1+list2)
print(array1+array2) # Adds the corresponding elements of the two arrays.
print(array1*2)
```

array indexing

```python
list1=[i+3 for i in range(50)]
print(list1)
array1=np.array(list1)
print(array1[0])
print(array1[-1])
print(array1[10:25:2])
print(array1[-1:-10:-1])
```

```python
arr=np.array([1,2,3,4,5])
try:
    print(arr[5])
except IndexError:
    print("Index out of bounds")
```

masking 

```python
print(arr[arr>2])
print(arr[arr%2==0])
list=[i for i in range(10)]
array=np.array(list)
print(array)
print(array[array%2!=0])
print(array[array%2==0])
array[array%2==0]=0
array[array%2!=0]=1
print(array)
print(array%2==0)
```

any => retruns boolen for given condition in the array true if any is true
all => true if all are true

```python
salary=np.random.randint(10000,50000,10)
maxsalary=40000
print(salary)
print(np.any(salary>maxsalary))
print(np.all(salary>maxsalary))
```

fancy indexing => access multiple index at same time  

```python
arr=np.random.randint(1,100,20)
print(arr)
#arr[2,-1,-6] error normal indexing
arr[[2,-1,-6]] #fancy indexing use [[]]
```

```python
arr=np.array([5,10,15,20,25,30,35,40,45,50])
print(arr)
print(arr[1::3]) #1. slice and print every third element from index 1
print(arr[-3::]) #2. last 4 element using negative indexing
print(arr[[2,4,6,8]]) #4. use fancy indexing to extract the values at positions [2,4,6,8]
arr[::2]=-1 #3. Replace the values at even indices with -1
print(arr)
```
