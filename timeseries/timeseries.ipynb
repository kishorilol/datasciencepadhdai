# Time Series

Usesd for forecasting

Date input column important

Forecasting ways:

ml model -> input with date and predict for 7 days

statistical -> ARIMA(Auto Regressive Integrated Moving Average), SARIMA(Seasonal), SARIMAX(With extended variable), Prophet(Overall Forecasting Model)

ARIMA -> Data with no seasonality, simple regular data

SARIMA -> if data has seasonal pattern, rainfall

SARIMAX -> Events with seasonal pattern, sales in october and november

Prophet -> handle every time series data

## Components 

1, Trend -> long term incerase or decrease

2, Seasonality -> repeatative pattern over time

3, Noise -> Outliers
