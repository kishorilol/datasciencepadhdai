```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

OHE (very necessary in NLP)
a column named online with true and false
ohe finds unique values 
new column for the no of unique value, 1 for true 1 for false
column name is set as <columnname>_<uniquevalues> here online_true and online_false
1 in online_true column for true and 1 in online_false column for false or else 0

it creates more number of columns, may sparse matrix (no of 0 > no of 1)

LE
checks unique values 
gives index to each unique value in ascending order
0 index for false and 1 index for true if false,true is the ascending order
then the column is 0 for false and 1 for true

doesnt create new column, no sparse matrix

if the order is large,medium,small
and the label are 0,1,2 this may fell bit confusing which can be modified in OE

OE
we can set label according to will using category=[small,medium,large]
we can control order

more unique values can make this method more hard to apply

Target Guided OE
