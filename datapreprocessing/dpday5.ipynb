Data Encoding -> changing categorical columns into numeric
techinques:
    OHE(One Hot Encoding)
    LE(Label Encoding)
    OE(Ordinal Encoding)
    Target Guided OE

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder, LabelEncoder, OrdinalEncoder
import seaborn as sns
```

OHE (very necessary in NLP)
a column named online with true and false
ohe finds unique values 
new column for the no of unique value, 1 for true 1 for false
column name is set as <columnname>_<uniquevalues> here online_true and online_false
1 in online_true column for true and 1 in online_false column for false or else 0

it creates more number of columns, may sparse matrix (no of 0 > no of 1)

```python
df=pd.DataFrame({
    'color': ['red', 'green', 'blue', 'green', 'red', 'blue']
})
df
```

```python
type(df)
```

```python
df.dtypes
```

```python
ohe=OneHotEncoder()
ohe
```

```python
ohe.fit_transform(df[['color']]).toarray()
```

```python
encoded=ohe.fit_transform(df[['color']]).toarray()
encoded
```

```python
ohe.get_feature_names_out(['color'])
```

```python
encoded_df=pd.DataFrame(
    encoded,columns=ohe.get_feature_names_out(['color'])
)
encoded_df
```

```python
pd.concat([df,encoded_df],axis=1)
```

LE
checks unique values 
gives index to each unique value in ascending order
0 index for false and 1 index for true if false,true is the ascending order
then the column is 0 for false and 1 for true

doesnt create new column, no sparse matrix

if the order is large,medium,small
and the label are 0,1,2 this may fell bit confusing which can be modified in OE

OE
we can set label according to will using category=[small,medium,large]
we can control order

more unique values can make this method more hard to apply

```python
df=pd.DataFrame({
    'color':['red','blue','green','green','blue','red']
})
df
```

```python
le=LabelEncoder()
le.fit_transform(df['color'])
```

```python
size=['small','small','small','large','large','medium']
le.fit_transform(size)
```

```python
df['color']=le.fit_transform(df['color'])
df
```

```python
df2=pd.DataFrame({
    'size':['small','small','small','large','large','medium']
})
df2
```

```python
encoder=OrdinalEncoder(categories=[['small','medium','large']])
```

```python
df2['size']=encoder.fit_transform(df2[['size']])
df2
```

Target Guided OE
we need target column
encoding a column based on target column using aggregate function if there are dublicate

```python
df=pd.DataFrame({
    'city': ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix', 'New York', 'Los Angeles', 'Chicago'],
    'price': [100, 200, 150, 300, 250, 120, 220, 180]
})
df
```

```python
df.groupby('city')
```

```python
df.groupby('city')['price']
```

```python
df.groupby('city')['price'].mean()
```

```python
type(df.groupby('city')['price'].mean())
```

```python
mean_price=df.groupby('city')['price'].mean().to_dict()
```

```python
df['city_encoded']=df['city'].map(mean_price)
df
```
