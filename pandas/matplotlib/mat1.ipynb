```python
import matplotlib.pyplot as plt
import numpy as np
x=np.linspace(0,10,200)
```

```python
y=np.sin(x)
z=np.cos(x)
```

plot gives line graph

```python
plt.plot(x,y,"--r")
plt.xlabel('this is x data')
plt.ylabel('this is y data')
plt.title('this is title')
plt.show() #middle ma show rakhena bhane duitai figure combine hunxa
plt.plot(x,z,"--y")
```

scatter plot

```python
x1=[1,2,3,4,5]
x2=[3,4,2,1,5]
x3=[2,5,2,4,1]
x4=[1,2,2,3,4]
plt.scatter(x1,x2,c='red')
plt.scatter(x3,x4,c='black',s=100,marker='o')
plt.xlabel('this is x')
plt.ylabel('this is y')
plt.title('this is title')
```

```python
x=np.random.rand(50)
y=np.random.rand(50)
plt.scatter(x,y)
plt.grid(axis="both")
```

```python
plt.scatter(x,y,c=np.random.rand(50),s=np.random.rand(50)*1000,cmap='winter',alpha=0.5)
```

cmap are set of colors

piechart

```python
label=["Apple","Samsung","Nokia","Nothing","Xiaomi"]
sizes=[20,40,30,25,10]
plt.pie(sizes,labels=label,autopct='%1.2f%%',explode=[0.2,0,0,0.1,0])
plt.title("pie chart")
```

3d flow: 
1. create blank figure => fig=plt.figure()
2. add subplot in blank figure fig.add_subplot()
3. keep the projection 3d =. fig.add_subplot(projection="3d")

```python
x=np.random.rand(50)
y=np.random.rand(50)
z=np.random.rand(50)
fig=plt.figure(figsize=(15,10))
ax=fig.add_subplot(projection="3d")
ax.plot(x,y,z,label="kishor") #for line graph
ax.scatter(x,y,z,label="kishor") # for scatter plot
ax.set_xlabel("x axis")
ax.set_ylabel("y axis")
ax.set_zlabel("z axis")
plt.legend(fontsize=10,loc="lower left",title="3d scatter plot")
```
