```python
import matplotlib.pyplot as plt
import numpy as np
x=np.linspace(0,10,200)
```

```python
y=np.sin(x)
z=np.cos(x)
```

plot gives line graph

```python
plt.plot(x,y,"--r")
plt.xlabel('this is x data')
plt.ylabel('this is y data')
plt.title('this is title')
plt.show() #middle ma show rakhena bhane duitai figure combine hunxa
plt.plot(x,z,"--y")
```

scatter plot

```python
x1=[1,2,3,4,5]
x2=[3,4,2,1,5]
x3=[2,5,2,4,1]
x4=[1,2,2,3,4]
plt.scatter(x1,x2,c='red')
plt.scatter(x3,x4,c='black',s=100,marker='o')
plt.xlabel('this is x')
plt.ylabel('this is y')
plt.title('this is title')
```

```python
x=np.random.rand(50)
y=np.random.rand(50)
plt.scatter(x,y)
plt.grid(axis="both")
```

```python
plt.scatter(x,y,c=np.random.rand(50),s=np.random.rand(50)*1000,cmap='winter',alpha=0.5)
```

cmap are set of colors 
alpha are used to control transperancy
marker is used to denote the shape the plots
