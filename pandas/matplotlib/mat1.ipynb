3d flow: 
1. create blank figure => fig=plt.figure()
2. add subplot in blank figure fig.add_subplot()
3. keep the projection 3d =. fig.add_subplot(projection="3d")

```python
x=np.random.rand(50)
y=np.random.rand(50)
z=np.random.rand(50)
fig=plt.figure(figsize=(15,10))
ax=fig.add_subplot(projection="3d")
ax.plot(x,y,z,label="kishor") #for line graph
ax.scatter(x,y,z,label="kishor") # for scatter plot
ax.set_xlabel("x axis")
ax.set_ylabel("y axis")
ax.set_zlabel("z axis")
plt.legend(fontsize=10,loc="lower left",title="3d scatter plot")
```

```python
days = ["Sunday","Monday","Tuesday","Wednesday","Thursday", "Friday"]
sleeping = [7.5, 6, 6, 11, 7, 5]
eating = [2, 3, 4, 3, 2, 3]
working = [2, 8, 7, 2, 2, 6]
playing = [8, 5, 7, 8, 13,11]
writing = [8, 5, 7, 8, 13, 17]
gaming = [-2, 5, 7, 8, 13, 13]
coding = [8, 5, 7, 8, 13, 4]

plt.stackplot(days,sleeping, eating, working, playing,writing,gaming,coding,
              labels=['Sleep', 'Eat', 'Work', 'Play', 'Write', 'Game', 'Code'],
              colors=['c','b','r','black','purple','orange','green'])
plt.legend(loc='upper left')
plt.xlabel('Days')
plt.xticks(rotation=90)
```
