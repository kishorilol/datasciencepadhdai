```python
iris = sns.load_dataset('iris')
iris.info()
```

```python
iris.isnull().sum()
```

```python
sns.lineplot(data=iris, x='sepal_length', y='sepal_width')
#or
plt.show()
sns.lineplot(data=iris, x=iris.sepal_length, y=iris.sepal_width)
```

```python
sns.scatterplot(x=iris['sepal_length'], y=iris['sepal_width'], hue=iris['species'],size=iris['petal_length'])
```

```python
iris.drop(columns=['species'], inplace=True)
iris.corr()
```

```python
sns.heatmap(iris.corr(), annot=True, cmap='winter',vmin=-1, vmax=1)
```

```python
sns.color_palette('winter', as_cmap=True)
```

```python
sns.palplot(sns.color_palette('viridis'))
```

displot makes distribution plot gives histogram in default and gives kde for kind=kde and gives kde with histogram if kde=true

```python
sns.displot(iris['sepal_length'],bins=20, color='green')
plt.show()
```

```python
sns.displot(iris['sepal_length'], kde=True, bins=10, color='green',rug=True)
```

```python
sns.displot(iris['sepal_length'], kind='kde', color='green')
plt.show()
```

```python
sns.kdeplot(data=iris['sepal_length'], color='green', shade=True)
```
