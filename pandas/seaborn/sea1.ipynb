displot makes distribution plot gives histogram in default and gives kde for kind=kde and gives kde with histogram if kde=true

```python
sns.displot(iris['sepal_length'],bins=20, color='green')
plt.show()
```

```python
sns.displot(iris['sepal_length'], kde=True, bins=10, color='green',rug=True)
```

```python
sns.displot(iris['sepal_length'], kind='kde', color='green')
plt.show()
```

```python
sns.kdeplot(data=iris['sepal_length'], color='green', shade=True)
```

```python
tips=sns.load_dataset('tips')
```

```python
tips
```

```python
tips['smoker'].value_counts()
```

```python
sns.scatterplot(x=tips.total_bill,y=tips.tip,hue=tips.time,style=tips.smoker)
```

relational plot => line plot and scatter plot both can be made by specifying in kind

```python
sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter', col='time')
```

catplot => bar, box, 

```python
sns.catplot(x='sex', y='total_bill', data=tips, kind='boxen', hue='smoker')
```

joint plot => gives joint plot scatter plot in middle and histogram at top and buttom

```python
sns.jointplot(x=tips.total_bill,y=tips.tip, kind='reg')
```

pair plot => we dont define x axis and y axis just data, only works on numerical columns and perform scatter plot if different columns and scatter plot if different columns

```python
iris.head()
```

```python
sns.pairplot(data=iris, kind='reg')
```
