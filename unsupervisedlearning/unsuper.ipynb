## Clustering

Grouping similar points

### K-Means Clustering

K : number of clusters

for K=3 

1, randomly pick k centroids

2, measure distance from centroids to every data points

3, cluster the centroid the data points near to the cluster

4, move the centroid

5, repeat until previous centroid = current centroid

This only works for circular clusters

how to know how many k 

use scatter plot to indentify

another is Elbow method has : WCSS, number of cluster, elbow point 
 elbow point defines how many clusters to make

### DBSCAN

core points -> points that has minimum or more pts in E neighbour

broder points -> point that has less than minimum points bust should have at least one core point in E neighbourhood

noise point -> neither core nor border points

Density Connected Points -> two points A and B are density connected if they are directly connected with many core points and the distance between two consecutive core point < E. two density connected points can be placed in same cluster
