## Clustering

Grouping similar points

### K-Means Clustering

K : number of clusters

for K=3 

1, randomly pick k centroids

2, measure distance from centroids to every data points

3, cluster the centroid the data points near to the cluster

4, move the centroid

5, repeat until previous centroid = current centroid

This only works for circular clusters

how to know how many k 

use scatter plot to indentify
